```python
# Importer les bibliothèques nécessaires
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Charger le fichier CSV
df = pd.read_csv('../data/steam_reviews.csv')


```

```python
# 2. Vérification de l'intégrité des données
# Afficher les premières lignes pour vérifier la structure du fichier
df.head()

# Vérifier les types de données
df.dtypes

# Vérifier les valeurs manquantes
df.isnull().sum()

# Vérifier les doublons
df.duplicated().sum()
```

```python
# 3. Nettoyage des données
# Supprimer les doublons
df = df.drop_duplicates()

# Gérer les valeurs manquantes - ici on supprime les lignes avec des valeurs manquantes
df = df.dropna()



# Vérifier les types après nettoyage
df.dtypes
```

```python
# Afficher les noms de colonnes pour vérifier
print(df.columns)
```

```python
# 4. Visualisation de la distribution des scores
df['weighted_vote_score'].hist(bins=10)
plt.title('Distribution des scores des joueurs')
plt.xlabel('Score')
plt.ylabel('Nombre d\'avis')
plt.show()
```
